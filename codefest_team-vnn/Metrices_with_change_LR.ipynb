```python
import glob, numpy as np, pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

def read_auto(p):
    with open(p, "r", encoding="utf-8", errors="ignore") as f:
        first = f.readline()
    return pd.read_csv(p, header=None) if "," in first else pd.read_csv(p, sep=r"\s+", header=None)

# Ground truth (second-last column)
y_true = read_auto("/content/drive/MyDrive/examples/test_data.txt").iloc[:, -2].astype(float).reset_index(drop=True)

rows = []
for pf in sorted(glob.glob("/content/drive/MyDrive/Lrchange/*")):
    y_pred = read_auto(pf).iloc[:, 0].astype(float).reset_index(drop=True)
    mae  = mean_absolute_error(y_true, y_pred)
    mse  = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2   = r2_score(y_true, y_pred)
    m    = y_true != 0
    mape = (np.abs((y_true[m] - y_pred[m]) / y_true[m]).mean() * 100) if m.any() else np.nan
    rows.append({"file": pf.split("/")[-1], "MAE": mae, "MSE": mse, "RMSE": rmse, "R2": r2, "MAPE_%": mape})

df = pd.DataFrame(rows)
print(df.round(6).to_string(index=False))
df.to_csv("lrchange_metrics.csv", index=False)
```
